## Time

```python
dates = ['created_date', 'closed_date']
df[dates] = df[dates].apply(pd.to_datetime)
```

```python
df['closed_date'].isnull().any()
# closed_date and status are related, when status is closed, there will be a closed date otherwise none.
```

```python
# Checking if there are any closed dates before open dates
any_closed_earlier = any(c < o for o, c in zip(df['created_date'], df['closed_date']))
print(any_closed_earlier)
```

```python
# CHecking for any negative times or impossible resolution times
df['resolution_time'] = df['closed_date'] - df['created_date']
has_negative = (df['resolution_time'].dt.total_seconds() < 0).any()
print("Negative resolution times:", has_negative)

has_long = (df['resolution_time'].dt.days > 1).any()

if has_long:
    print("Unusually long resolution time detected")
```

```python
print(df['resolution_time'].describe())
```

## Coordinates

```python
# Checking nulls in latitude and longitude. Also if any coordinates go outside NYC
print("Missing latitude:", df['latitude'].isna().sum())
print("Missing longitude:", df['longitude'].isna().sum())
```

```python
# Checking coords outside
outside_nyc = df[
    (df['latitude'] < 40.49) |
    (df['latitude'] > 40.92) |
    (df['longitude'] < -74.27) |
    (df['longitude'] > -73.68)
]

print("Coordinates outside NYC:", len(outside_nyc))
```

```python
# Checking for invalid coords
invalid_coordinates = df[
    (df['latitude'].notna()) &
    (df['longitude'].notna()) &
    (
        (df['latitude'] < -90) |
        (df['latitude'] > 90) |
        (df['longitude'] < -180) |
        (df['longitude'] > 180)
    )
]

print("Invalid coordinates:", len(invalid_coordinates))
```

## Categorical Variables and Values

```python
# Checking categorical variables
categorical_columns = ['borough', 'status', 'agency']

for col in categorical_columns:
    print(f"\n--- {col.upper()} ---")
    print(df[col].value_counts(dropna=False))
```

```python
# Checking for top 20 complaints
print(df['complaint_type'].value_counts().head(20))
```

```python
# Checking for unspecified values
for col in categorical_columns + ['complaint_type']:
    print(f"\n--- Checking {col} ---")

    unspecified = df[col].astype('string').str.strip().str.lower()

    print("Unspecified:", unspecified.eq('unspecified').sum())
    print("Blank:", unspecified.eq('').sum())
    print("Null:", df[col].isna().sum())
```

```python
for col in categorical_columns + ['complaint_type']:
    casing = df[col].dropna().astype(str).str.strip()

    print(f"\n{col} - Unique values:", casing.nunique())
    print(casing.value_counts().head(20))
```
